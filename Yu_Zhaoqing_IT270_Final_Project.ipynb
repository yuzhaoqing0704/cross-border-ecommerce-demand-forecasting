# 6. Feature Selection and Target Variable

```python
features = [
    'Product_Category',
    'Brand',
    'Region',
    'Country',
    'Marketing_Spend',
    'Discount_Rate',
    'Product_Price',
    'Competitor_Price',
    'Website_Traffic',
    'Conversion_Rate',
    'Search_Trend_Index',
    'Festival_Season',
    'Weather_Index',
    'Currency_Exchange_Rate',
    'Tariff_Rate',
    'Fuel_Cost_Index',
    'Port_Congestion_Index',
    'Lead_Time_Days',
    'Supplier_Reliability',
    'Return_Rate',
    'Historical_Sales',
    'Inventory_Level',
    'Shipping_Mode',
    'Shipping_Cost',
    'Delivery_Time_OnTime',
    'Service_Level',
    'Total_Logistics_Cost'
]

target = 'Demand_Next_Period'
```

# 7. Data Preprocessing

```python
X_encoded = pd.get_dummies(X, drop_first=True)
```

### Data Preprocessing Summary

In this stage, categorical variables were converted into machine learning compatible numerical variables using one-hot encoding. This preprocessing step allows machine learning algorithms to interpret categorical business variables such as product category, country, brand, region, and shipping mode.

After preprocessing, the dataset expanded from the original selected features into 50 machine learning features. The processed dataset is now fully prepared for machine learning model development and predictive demand forecasting analysis.

This step is important because machine learning models require numerical input variables in order to identify patterns and relationships associated with future product demand in cross-border trade operations.

# 8. Train-Validation-Holdout Split


```python
from sklearn.model_selection import train_test_split

# Step 1: Split into Training Set (50%) and Temporary Set (50%)
X_train, X_temp, y_train, y_temp = train_test_split(
    X_encoded,
    y,
    test_size=0.5,
    random_state=1
)

# Step 2: Split Temporary Set into Validation Set (30%) and Holdout Set (20%)
X_validation, X_holdout, y_validation, y_holdout = train_test_split(
    X_temp,
    y_temp,
    test_size=0.4,
    random_state=1
)

print(X_train.shape)
print(X_validation.shape)
print(X_holdout.shape)

print(y_train.shape)
print(y_validation.shape)
print(y_holdout.shape)
```

### Train-Validation-Holdout Split Summary

The dataset was divided into training, validation, and holdout datasets using a 50% / 30% / 20% split ratio.

The training dataset contains 12,500 observations and is used to train the machine learning models. The validation dataset contains 7,500 observations and is used for model tuning and model comparison. The holdout dataset contains 5,000 observations and is reserved for final model evaluation on unseen data.

This approach helps reduce overfitting and provides a more reliable evaluation of predictive performance. Using separate training, validation, and holdout datasets is an important step in machine learning and demand forecasting analysis.

# 9. Naive Baseline Model


```python
# Naive benchmark model: always predict the average demand from the training set
baseline_pred_validation = np.repeat(y_train.mean(), len(y_validation))

baseline_rmse = np.sqrt(mean_squared_error(y_validation, baseline_pred_validation))
baseline_mae = mean_absolute_error(y_validation, baseline_pred_validation)
baseline_r2 = r2_score(y_validation, baseline_pred_validation)

print("Baseline Validation Results")
print("RMSE:", baseline_rmse)
print("MAE:", baseline_mae)
print("R²:", baseline_r2)
```

### Naive Baseline Model Summary

A naive baseline model was created by predicting the average future demand value from the training dataset for every observation in the validation dataset. This model does not use any predictor variables.

The purpose of the baseline model is to provide a simple performance benchmark. The Linear Regression and Decision Tree Regression models should produce lower RMSE and MAE values than this baseline to show that the predictors add useful forecasting information.

# 10. Linear Regression Model


```python
lr = LinearRegression()
lr.fit(X_train, y_train)

lr_validation_pred = lr.predict(X_validation)

lr_rmse = np.sqrt(mean_squared_error(y_validation, lr_validation_pred))
lr_mae = mean_absolute_error(y_validation, lr_validation_pred)
lr_r2 = r2_score(y_validation, lr_validation_pred)

print("Linear Regression Validation Results")
print("RMSE:", lr_rmse)
print("MAE:", lr_mae)
print("R²:", lr_r2)
```

### Linear Regression Model Summary

The Linear Regression model was trained on the training dataset and evaluated on the validation dataset. RMSE and MAE measure prediction error, while R² shows how much variation in future demand is explained by the model.

These validation results are compared with the baseline model and the Decision Tree Regression model to determine whether a more complex model improves demand forecasting performance.


# 11. Decision Tree Regression Model and Tuning


```python
depth_results = []

for depth in [3, 5, 7, 10, 15, 20]:
    dt_candidate = DecisionTreeRegressor(max_depth=depth, random_state=1)
    dt_candidate.fit(X_train, y_train)
    validation_pred = dt_candidate.predict(X_validation)
    
    depth_results.append({
        "max_depth": depth,
        "RMSE": np.sqrt(mean_squared_error(y_validation, validation_pred)),
        "MAE": mean_absolute_error(y_validation, validation_pred),
        "R2": r2_score(y_validation, validation_pred)
    })

depth_results_df = pd.DataFrame(depth_results)
depth_results_df
```

```python
best_depth = depth_results_df.loc[depth_results_df["RMSE"].idxmin(), "max_depth"]

dt = DecisionTreeRegressor(max_depth=int(best_depth), random_state=1)
dt.fit(X_train, y_train)

dt_validation_pred = dt.predict(X_validation)

dt_rmse = np.sqrt(mean_squared_error(y_validation, dt_validation_pred))
dt_mae = mean_absolute_error(y_validation, dt_validation_pred)
dt_r2 = r2_score(y_validation, dt_validation_pred)

print("Best max_depth:", best_depth)
print("Decision Tree Validation Results")
print("RMSE:", dt_rmse)
print("MAE:", dt_mae)
print("R²:", dt_r2)
```

### Decision Tree Regression Summary

The Decision Tree Regression model was tuned by testing several `max_depth` values on the validation set. This step helps control overfitting and provides a fairer comparison with the Linear Regression model.

The final Decision Tree model uses the depth with the lowest validation RMSE.


# 12. Validation Model Comparison


```python
model_comparison = pd.DataFrame({
    "Model": ["Naive Baseline", "Linear Regression", "Decision Tree Regression"],
    "Validation_RMSE": [baseline_rmse, lr_rmse, dt_rmse],
    "Validation_MAE": [baseline_mae, lr_mae, dt_mae],
    "Validation_R2": [baseline_r2, lr_r2, dt_r2]
})

model_comparison_sorted = model_comparison.sort_values(by="Validation_RMSE")
model_comparison_sorted
```

### Validation Model Comparison Analysis

The models were compared using the validation dataset. The main selection metric is RMSE because the project goal is to forecast future demand, and RMSE directly measures the size of prediction errors.

The naive baseline is included as the minimum performance benchmark. A useful forecasting model should outperform this simple average-demand prediction. After comparing validation RMSE, MAE, and R², the model with the lowest validation RMSE is selected as the final model.

The holdout dataset is not used during model selection. It is reserved for the final evaluation only.

# 13. Final Holdout Evaluation


```python
# Select the best model based on validation RMSE
best_model_name = model_comparison.loc[model_comparison["Validation_RMSE"].idxmin(), "Model"]

if best_model_name == "Linear Regression":
    final_model = lr
elif best_model_name == "Decision Tree Regression":
    final_model = dt
else:
    final_model = None

print("Selected best model based on validation RMSE:", best_model_name)
```

```python
if final_model is not None:
    holdout_pred = final_model.predict(X_holdout)
else:
    holdout_pred = np.repeat(y_train.mean(), len(y_holdout))

holdout_rmse = np.sqrt(mean_squared_error(y_holdout, holdout_pred))
holdout_mae = mean_absolute_error(y_holdout, holdout_pred)
holdout_r2 = r2_score(y_holdout, holdout_pred)

holdout_results = pd.DataFrame({
    "Final_Model": [best_model_name],
    "Holdout_RMSE": [holdout_rmse],
    "Holdout_MAE": [holdout_mae],
    "Holdout_R2": [holdout_r2]
})

holdout_results
```

### Final Holdout Evaluation Summary

After the best model was selected using validation RMSE, the holdout dataset was used for the final evaluation. The holdout data represents unseen observations that were not used for training, tuning, or model selection.

This step follows the correct machine learning workflow: train the models on the training data, compare and tune models on the validation data, and evaluate the selected final model once on the holdout data. The holdout results represent the final predictive performance of this project.

# 14. Model Visualization


```python
plt.figure(figsize=(8,6))

plt.scatter(y_holdout, holdout_pred, alpha=0.5)

plt.xlabel("Actual Demand")
plt.ylabel("Predicted Demand")
plt.title("Actual vs Predicted Future Demand on Holdout Set")

plt.show()
```

```python
residuals = y_holdout - holdout_pred

plt.figure(figsize=(8,6))
plt.scatter(holdout_pred, residuals, alpha=0.5)
plt.axhline(y=0, linestyle='--')

plt.xlabel("Predicted Demand")
plt.ylabel("Residuals")
plt.title("Residual Plot on Holdout Set")

plt.show()
```

### Actual vs Predicted and Residual Plot Summary

The actual vs. predicted plot compares the model's final predictions with the real demand values in the holdout dataset. Points closer to a diagonal pattern indicate stronger predictive performance.

The residual plot shows the prediction errors. A better model should have residuals spread around zero without a strong pattern. This helps evaluate whether the model is systematically overpredicting or underpredicting demand.
